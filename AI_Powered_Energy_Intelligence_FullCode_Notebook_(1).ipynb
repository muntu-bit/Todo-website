# **Problem Statement**

## **Business Context**

Lumina Energy Partners is a premier global investment firm managing a portfolio across traditional fossil fuels, renewable power generation, and advanced energy infrastructure. The global energy landscape is currently undergoing an unprecedented and highly complex transition. As the “Age of Electricity” takes hold, power demand is surging due to the electrification of transport and heating and the rapid expansion of AI-driven data centres. At the same time, the market is experiencing major structural shifts, including a coming wave of liquefied natural gas production, plateauing global oil demand, slower coal plant retirements due to policy support, and persistent bottlenecks in power grid infrastructure that constrain renewable integration.

To maximise portfolio returns and mitigate risk, Lumina’s investment committee requires rapid cross-sector intelligence. Evaluating an investment in a US natural gas facility, for example, requires analysts to synthesise insights spanning multiple energy markets. They must understand how AI data centres could account for almost half of projected US electricity demand growth and potentially add nearly 20 bcm of annual natural gas demand by 2030. At the same time, they must consider countervailing developments such as the slowdown in coal plant retirements due to federal policy support and the growing constraints on renewable deployment caused by grid congestion. These interdependencies require analysts to combine signals from multiple technical sources to produce a coherent macro outlook.

Currently, Lumina’s analysts rely on manual review of dense and highly technical International Energy Agency reports covering Coal, Oil, Gas, Electricity, and Renewables. Preparing a cross-sector investment briefing typically requires two to three days of work, during which analysts review five to ten separate reports and spend roughly 65% of their time locating relevant passages rather than performing financial modelling or scenario analysis. The process relies heavily on manual search across thousands of pages of PDFs and cross-referencing fragments of information between reports.

This workflow creates measurable operational inefficiencies and analytical risk. Citation traceability is inconsistent, subtle correlations between reports may be overlooked, and valuable analyst time is diverted away from high-value modelling and strategic evaluation. In fast-moving market conditions or competitive bidding windows, the delay inherent in manual synthesis can mean that investment opportunities are missed before a complete briefing is produced.

To remain competitive, Lumina requires an AI-assisted system capable of reducing briefing preparation time by at least 90%, retrieving relevant report passages with at least 90% accuracy, and generating fully grounded responses with complete source attribution and minimal hallucination risk. By transitioning from manual document review to a Retrieval Augmented Generation system grounded exclusively in authoritative IEA reports, Lumina can enable analysts to ask complex natural language questions and receive instant, synthesised, and fully cited strategic briefings. This approach would eliminate days of manual document review, reduce the risk of human oversight, and allow analysts to focus on financial modelling and strategic decision-making rather than time-consuming information retrieval.

## **Objective**

Your objective as a Data Scientist is to design, build, and evaluate a prototype Retrieval Augmented Generation (RAG) pipeline that enables Lumina’s analysts to rapidly obtain accurate and fully traceable energy market insights from a corpus of dense, technical International Energy Agency market reports. The system must synthesise information across multiple reports while producing high-fidelity, hallucination-free responses suitable for high-stakes investment decision-making.

Key tasks include:
- **Data Ingestion and Preprocessing**: Extract text, quantitative data points, and complex policy details from the provided IEA PDF reports. Implement an intelligent chunking strategy that preserves the context of dense market analyses, outlook sections, and policy frameworks so that cross-sector insights can be retrieved accurately.
- **Vector Embedding and Storage**: Select an embedding model capable of capturing the semantic meaning of highly technical energy terminology and relationships across the reports. Generate semantic representations of the text and store these embeddings in a vector database to enable efficient similarity search across the entire corpus.
- **Advanced Retrieval**: Develop a robust retrieval mechanism capable of surfacing the most relevant passages across multiple reports in response to complex, multi-part analyst queries. The retrieval layer should support cross-sector questions that connect developments such as electricity demand growth, LNG supply expansion, and coal plant utilisation.
- **LLM Synthesis and Citation**: Integrate a Large Language Model to synthesise retrieved evidence into coherent, executive-level investment briefings. The generated responses must explicitly cite the source reports and relevant sections or pages for each key claim in order to ensure auditability and minimise hallucination risk.
- **Evaluation**: Establish evaluation metrics to assess system performance against benchmark questions generated by Lumina’s senior energy analysts. Evaluation should measure retrieval relevance and factual accuracy, to achieve approximately 90% retrieval accuracy and significantly reduce the time required to prepare cross-sector investment briefings.

```python
# Load the JSON file and extract values, handling potential RTF formatting wrapper
import json
import re

file_name = 'config.json'
config = {}

try:
    with open(file_name, 'r', encoding='utf-8') as file:
        content = file.read()

    # Check if the file is RTF formatted
    if 'rtf' in content or content.strip().startswith('{\\rtf') or '\\pard' in content:
        print("Warning: 'config.json' is saved as an RTF (Rich Text) file.")
        print("Extracting plain text contents...")

        # Clean up RTF formatting safely using standard single-line string replacements
        cleaned = content.replace('\\\\\n', '').replace('\\\n', '').replace('\\n', '')
        # Remove RTF control words like \pard, \cf0, \partightenfactor0, etc.
        cleaned = re.sub(r'\\[A-Za-z0-9]+', '', cleaned)
        # Clean up stray backslashes
        cleaned = cleaned.replace('\\', '')

        # Extract the JSON block {...}
        json_match = re.search(r'\s*\{[^{}]*"API_KEY"[^}]*\}\s*', cleaned)
        if json_match:
            content = json_match.group(0).strip()
        else:
            content = cleaned.strip()

    # Attempt to parse the cleaned JSON content
    try:
        config = json.loads(content)
    except json.JSONDecodeError:
        # Clean up potential leading/trailing RTF brackets if any remain
        content = content.strip().strip('{}')
        # Reconstruct clean JSON using regex extraction
        api_key_match = re.search(r'"API_KEY"\s*:\s*"([^"]+)"', content)
        api_base_match = re.search(r'"OPENAI_API_BASE"\s*:\s*"([^"]+)"', content)
        if api_key_match:
            config["API_KEY"] = api_key_match.group(1)
        if api_base_match:
            config["OPENAI_API_BASE"] = api_base_match.group(1)

except Exception as e:
    print(f"Failed to parse config file: {e}")
    print("Please recreate 'config.json' as a plain text file, not Rich Text.")

# Extract the OpenAI API key and API base from config
OPENAI_API_KEY = config.get("API_KEY")
OPENAI_API_BASE = config.get("OPENAI_API_BASE")

if OPENAI_API_KEY:
    print("API Key successfully loaded!")
else:
    print("Could not extract API Key. Please verify 'config.json' format.")
```

## **Define the response function**

```python
def generate_response(prompt: str, system_prompt: str = "You are an expert energy intelligence analyst providing accurate, detailed, and well-reasoned insights based on industry data and reports.") -> str:
    """
    Generates a response using the OpenAI client configured in the notebook.
    """
    try:
        # Using the OpenAI client initialization
        client = OpenAI(
            api_key=OPENAI_API_KEY,
            base_url=OPENAI_API_BASE
        )

        response = client.chat.completions.create(
            model="gpt-4o",  # Or your specified deployment/model name
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": prompt}
            ],
            temperature=0.2
        )

        return response.choices[0].message.content.strip()

    except Exception as e:
        return f"Error generating response: {e}"

# Example test run of the function
print("Response function defined successfully!")
```

### Query 1: How is the rapid global expansion of artificial intelligence data centres impacting overall electricity demand and straining existing power grid infrastructure?

AI data centres are becoming a major driver of electricity demand growth, requiring large amounts of continuous power and increasing pressure on existing energy infrastructure. Their rapid deployment often outpaces the expansion of transmission and distribution networks, leading to localized grid congestion and reliability challenges.

Growing data centre demand may also contribute to higher electricity prices and increased market volatility in regions with limited grid capacity. To address these challenges, stakeholders are investing in grid upgrades, battery energy storage, and co-location of data centres with low-carbon energy sources such as renewable and nuclear power.

### Query 2: How is the unprecedented wave of new US liquefied natural gas (LNG) export capacity expected to impact natural gas affordability and spur additional demand in price-sensitive Asian markets by 2030?

The rapid expansion of U.S. LNG export capacity is expected to strengthen the country's role in global natural gas markets, supplying a growing share of international demand by 2030. While increased exports support domestic production and energy trade, they may also place upward pressure on domestic natural gas prices, affecting affordability for U.S. consumers, utilities, and industrial users.

Globally, the addition of new LNG export projects is expected to increase supply and moderate international gas prices. Lower LNG prices could improve energy affordability in Asia, particularly in emerging economies where cost has been a major barrier to natural gas adoption. This may encourage greater use of natural gas as a transition fuel alongside renewable energy sources.

### Query 3: How are the surge in US electricity demand and the 2025 federal emergency policy interventions collectively affecting the retirement schedules, capacity planning, and generation output of domestic coal-fired power plants?

Rising electricity demand and federal reliability measures have slowed the retirement of U.S. coal-fired power plants. Several planned closures were postponed or canceled, and emergency orders from the U.S. Department of Energy kept some facilities available to support grid reliability.

While these actions have extended the operation of certain coal plants, many now serve primarily as backup capacity during periods of high demand rather than as continuous baseload generators.

### Query 4:  How are the increasing frequency of negative wholesale electricity prices and the regulatory shift towards two-sided Contracts for Difference (CfDs) in Europe altering the revenue expectations and financial agility of developers investing in utility-scale solar PV?

The rapid growth of utility-scale solar PV across Europe has increased the frequency of negative electricity prices, particularly during peak solar production hours. This trend reduces solar capture prices and creates revenue challenges for merchant solar projects.

To reduce investment risk, the European Commission's electricity market reforms have expanded the use of two-sided Contracts for Difference (CfDs), which provide revenue stability by compensating developers when market prices fall below a predetermined strike price while requiring repayment when prices exceed it. This lowers financing costs and supports continued renewable energy investment.

At the same time, developers are increasingly incorporating battery energy storage systems (BESS) to shift generation away from low-value or negative-price periods, improving project revenues and operational flexibility.

### Query 5: How do the tax credit modifications under the US "One Big Beautiful Bill Act" (OBBBA) affect the investment economics of using domestic versus imported feedstocks for Sustainable Aviation Fuel (SAF), and what cascading impact will this biofuel transition have on the capacity rationalisation of traditional US West Coast refineries?

The One Big Beautiful Bill Act (OBBBA) modified the Section 45Z Clean Fuel Production Credit by restricting eligible feedstocks to those produced in the United States, Canada, or Mexico, increasing reliance on North American supply chains for sustainable aviation fuel (SAF) and renewable diesel production.

These changes are expected to increase competition for domestic feedstocks such as soybean oil and tallow, raising input costs while maintaining incentives for clean fuel investment through the extension of the 45Z credit.

For West Coast refineries, stricter feedstock requirements and regional decarbonization policies are accelerating the shift toward renewable fuel production, facility conversion, or capacity reductions.

Loading the extracted IEA PDF reports

```python
from pathlib import Path
from langchain_community.document_loaders import PyMuPDFLoader

# Folder containing extracted IEA reports
folder_path = Path("/content/IEAReports")

# Find all PDF files, including those in subfolders
pdf_files = sorted(folder_path.rglob("*.pdf"))

print(f"Number of PDF files found: {len(pdf_files)}")

# Load the PDFs using PyMuPDFLoader to bypass pypdf stream token limits
documents = []

for pdf_file in pdf_files:
    print(f"Loading: {pdf_file.name}")

    loader = PyMuPDFLoader(str(pdf_file))
    pages = loader.load()

    documents.extend(pages)

print(f"Total pages loaded: {len(documents)}")
```

Split the documents into chunks

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Configure the text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    length_function=len,
    separators=["\n\n", "\n", ". ", " ", ""]
)

# Split the loaded documents
chunks = text_splitter.split_documents(documents)

print("Total chunks created:", len(chunks))
```

# **Question Answering using RAG**

## **Define the RAG response function**

```python
rag_system_prompt = """
You are an expert energy market research analyst
specializing in oil refining, sustainable aviation fuel,
biofuels, energy transition, and global energy policy.

Your task is to answer research questions using
the retrieved context from IEA reports.

Follow these instructions:

1. Use the retrieved context as your primary evidence.
2. Provide accurate, relevant, and comprehensive answers.
3. Do not invent facts, statistics, or policy details.
4. If the context is insufficient, explicitly state
   what information is missing.
5. Distinguish established facts from interpretations.
6. Explain economic, regulatory, and market implications.
7. Support factual claims with source references
   when available.
8. Organize responses logically and professionally.
9. Avoid unnecessary repetition.

Your goal is to generate reliable, evidence-based
research answers grounded in the supplied documents.
"""

print("RAG system prompt defined successfully!")
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)

print("RAG retriever created successfully!")
```

Define RAG response function

```python
def generate_rag_response(query):

    # Step 1: Retrieve relevant documents
    retrieved_docs = retriever.invoke(query)

    # Step 2: Combine retrieved documents into context
    context_parts = []

    for i, doc in enumerate(retrieved_docs, start=1):

        source = doc.metadata.get("source", "Unknown")
        page = doc.metadata.get("page", 0) + 1

        context_parts.append(
            f"[Source {i}: {source}, Page {page}]\n"
            f"{doc.page_content}"
        )

    context = "\n\n".join(context_parts)

    # Step 3: Create the user prompt
    user_prompt = f"""
Answer the following research question using
the retrieved IEA report context.

Research Question:
{query}

Retrieved Context:
{context}

Provide a well-structured, evidence-based answer.
Cite relevant source numbers and page numbers.
If the context does not contain enough information,
state that limitation clearly.
"""

    # Step 4: Generate the LLM response
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": rag_system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],
        temperature=0.2
    )

    # Step 5: Return the generated answer
    return response.choices[0].message.content


print("RAG response function defined successfully!")
```
